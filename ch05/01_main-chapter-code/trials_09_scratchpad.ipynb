```python
import torch

t = torch.arange(10)
print("t            :", t)
print("t[3]         :", t[3].item())          # single element
print("t[2:5]       :", t[2:5])                # slice

assert t[3].item() == 3
assert torch.equal(t[2:5], torch.tensor([2, 3, 4]))
print("OK: plain indexing/slicing behave as expected")
```
